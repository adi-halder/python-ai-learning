```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, SystemMessage

# Connect to Groq
llm = ChatGroq(
    api_key="YOUR_API_KEY",
    model="llama-3.3-70b-versatile"
)

print("LangChain connected!")
```

```python
# Send a message to the AI
messages = [
    SystemMessage(content="you are a helpful assistant."),
    HumanMessage(content="Give me one tip to stay consistent while learning to code."),
]

response = llm.invoke(messages)
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

# Create a reusable prompt template
template = ChatPromptTemplate.from_messages([
    ("system", "You are an expert {subject} tutor for BTech students."),
    ("human", "Explain {topic} in simple terms with a real life example.")
])

# Use the template with different subjects and topics
prompt1 = template.invoke({
    "subject": "Machine Learning",
    "topic": "Neural Networks"
})

prompt2 = template.invoke({
    "subject": "Data Structures",
    "topic": "Binary Search Trees"
})

response1 = llm.invoke(prompt1)
response2 = llm.invoke(prompt2)

print("=== ML Explanation ===")
print(response1.content)
print("\n=== DSA Explanation ===")
print(response2.content)
```

```python
from langchain_core.output_parsers import StrOutputParser

# Create a chain — template | llm | parser
chain = template | llm | StrOutputParser()

# Run the chain in one line!
result = chain.invoke({
    "subject": "Python Programming",
    "topic": "Recursion"
})

print(result)
```

```python
result = chain.invoke({"subject": "Operating Systems", "topic": "FCFS"})
print(result)
```

```python
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

# Store for conversation history
store = {}

def get_session_history(session_id):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

# Chain with memory
chain_with_memory = RunnableWithMessageHistory(
    llm,
    get_session_history
)

# First message
response1 = chain_with_memory.invoke(
    [HumanMessage(content="My name is Aditya and I am learning AI")],
    config={"configurable": {"session_id": "aditya123"}}
)
print("AI:", response1.content)

# Second message — does it remember?
response2 = chain_with_memory.invoke(
    [HumanMessage(content="What is my name and what am I learning?")],
    config={"configurable": {"session_id": "aditya123"}}
)
print("AI:", response2.content)
```
