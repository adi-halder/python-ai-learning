```python
student = {
    "name": "Aditya Halder",
    "roll": 42,
    "marks": {
        "Maths": 91,
        "Physics": 92,
        "Python": 78,
        "Chemistry": 88
    }
}
student["marks"]["OS"] = 84

print("===== REPORT CARD =====")
print("Name    :", student["name"])

print("Roll No :", student["roll"])
print("-----------------------")
for subject, mark in student["marks"].items():
    if mark >= 90:
        print(subject, "->", mark, "Excellent")
    else:
        print(subject, "->", mark, "Good")
print("-----------------------")
marks_list = list(student["marks"].values())
print("Total   :", sum(marks_list))
print("Average :", sum(marks_list) / len(marks_list))
print("Highest :", max(marks_list))
print("Lowest  :", min(marks_list))
print("=======================")
```
