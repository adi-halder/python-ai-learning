```python
import pandas as pd
```

```python
data = {
    "Name" : ["John", "Aditya", "Tufail", "Faiez", "Purvi"],
    "Marks" : [85, 92, 65, 84, 77],
    "Age" : [20, 20, 21, 22, 27],
    "City" : ["Dehli", "Mumbai", "Gurgaon", "Pune", "Delhi"],
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head(3))
print(df.tail(2))
print(df.shape)
```

```python
print(df["Marks"])
print(df["Name"][0])
print(df[df["Marks"] > 80])
```

```python
df["Grade"] = df["Marks"].apply(lambda x : "A" if x >= 90 else "B" if x >=75 else "C")
print(df)
```

```python
# Load a real dataset - Titanic survival data
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)
print(titanic.shape)
print(titanic.head())
```

```python
print(titanic["Survived"].value_counts())


print("Average Age:", titanic["Age"].mean())


print(titanic.isnull().sum())
```

```python
print(titanic.groupby("Sex")["Survived"].mean())
print(titanic.groupby("Pclass")["Survived"].mean())
```

```python
import matplotlib.pyplot as plt

titanic["Survived"].value_counts().plot(kind="bar", color=["red", "green"])
plt.title("Titanic Survival Count")
plt.xlabel("0 = Died, 1 = Survived")
plt.ylabel("Number Of Passengers")
plt.show()
```
