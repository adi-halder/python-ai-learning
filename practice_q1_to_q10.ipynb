```
numbers = [34, 67, 23, 89, 45, 12]

def find_highest(numbers):
  highest = numbers[0]

  for number in numbers:
    if number > highest:
      highest = number

  return highest

print(find_highest(numbers))
```

```
# 1. dictionary first
students = {
    "Aditya": 85,
    "Riya": 92,
    "Arjun": 67,
    "Priya": 78,
    "Rohan": 55
}

def top_student(students):
    top_name = list(students.keys())[0]
    top_mark = list(students.items())[0][1]

    for student, mark in students.items():
        if mark > top_mark:
            top_mark = mark
            top_name = student

    return top_name

print(top_student(students))
```

```
students = {
    "Aditya": 85,
    "Riya": 92,
    "Arjun": 67,
    "Priya": 78,
    "Rohan": 55
}

def top_student(students):
    top_name = list(students.keys())[0]
    top_mark = list(students.values())[0]

    for student, mark in students.items():
        if mark > top_mark:
            top_mark = mark
            top_name = student

    return top_name

print(top_student(students))
```
