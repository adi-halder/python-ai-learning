```python
# Connect to Groq AI
client = Groq(api_key="YOUR_API_KEY_HERE")

print("Connected to AI!")
```

```python
# PlanIT AI Brain

def generate_plan(goal, hours, level):
    prompt = f"""
    You are PlanIT, an AI daily planner for BTech students.

    Student details:
    - Goal: {goal}
    - Available hours today: {hours}
    - Current level: {level}

    Generate a focused daily plan with exactly 4 tasks.
    Keep each task specific, actionable and achievable in the given hours.
    Format exactly like this:

    🎯 YOUR PLAN FOR TODAY
    ----------------------
    1. [task] (X mins)
    2. [task] (X mins)
    3. [task] (X mins)
    4. [task] (X mins)

    💡 TIP: [one motivational tip for the student]
    """

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}]
    )

    return response.choices[0].message.content

# Test it!
plan = generate_plan(
    goal="Get an AI internship",
    hours=4,
    level="Beginner"
)

print(plan)
```

```python
def generate_plan(goal, hours, level, year):
    prompt = f"""
    You are PlanIT, an AI mentor and daily planner for BTech students.
    You don't just give tasks — you guide, teach and roadmap like a senior mentor.

    Student profile:
    - Goal: {goal}
    - BTech Year: {year}
    - Available hours today: {hours}
    - Current level: {level}

    Generate a smart daily plan with exactly 4 tasks.
    For EACH task provide:
    - The task with time
    - WHY it matters for their goal
    - HOW to do it (specific steps, resources, platforms)
    - What to do TOMORROW to continue

    Format exactly like this:

    🎯 PLANIT — YOUR DAILY ROADMAP
    ================================

    TASK 1: [task name] ([X] mins)
    📌 Why: [why this matters for their specific goal]
    📖 How: [exact steps — platform, what to study, how to practice]
    ➡️ Tomorrow: [what to do next]

    TASK 2: [task name] ([X] mins)
    📌 Why: [why this matters]
    📖 How: [exact steps]
    ➡️ Tomorrow: [next step]

    TASK 3: [task name] ([X] mins)
    📌 Why: [why this matters]
    📖 How: [exact steps]
    ➡️ Tomorrow: [next step]

    TASK 4: [task name] ([X] mins)
    📌 Why: [why this matters]
    📖 How: [exact steps]
    ➡️ Tomorrow: [next step]

    ================================
    💡 MENTOR TIP: [personalized advice based on their year and goal]
    🔥 MOTIVATION: [one line that hits hard]
    """

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=1500
    )

    return response.choices[0].message.content

# Test with YOUR profile!
plan = generate_plan(
    goal="Learn AI, get an AI internship",
    hours=5,
    level="Beginner",
    year="2nd year"
)
print(plan)
```
